## Point Processing of Image

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt

img = cv2.imread(r'C:\Users\Debasish Das\Desktop\Computer-Vision-Resources\dataset\cup.jpg', cv2.IMREAD_GRAYSCALE)
# Check image
if img is None:
    print("Image not found!")
    exit()
```

```python
# Image shape
# length
width=len(img)
hight=len(img[0])
print("Width",width)
print("hight",hight)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# -------------------------------
# 1. Image Negative
# -------------------------------
negative = 255 - img

# -------------------------------
# 2. Brightness Increase
# -------------------------------
brightness = np.clip(img + 50, 0, 255).astype(np.uint8)

# -------------------------------
# 3. Brightness Decrease
# -------------------------------
dark = np.clip(img - 50, 0, 255).astype(np.uint8)

# -------------------------------
# 4. Thresholding
# -------------------------------
threshold = np.where(img > 128, 255, 0).astype(np.uint8)

# -------------------------------
# 5. Contrast Stretching
# -------------------------------
contrast = cv2.normalize(
    img,
    None,
    alpha=0,
    beta=255,
    norm_type=cv2.NORM_MINMAX
)

# -------------------------------
# Display Results
# -------------------------------
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(brightness, cmap="gray")
plt.title("Brightness +50")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(dark, cmap="gray")
plt.title("Brightness -50")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(threshold, cmap="gray")
plt.title("Threshold")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.tight_layout()
plt.show()
```
